## INTRODUCTION

In modern supply chain operations, efficient shipping is critical to maintaining customer satisfaction and controlling logistics costs. However, many organizations lack clear visibility into how their factory-to-customer delivery routes perform over time. This results in reactive decision-making, where delays and inefficiencies are addressed only after they occur rather than being proactively managed.

This project focuses on analyzing shipping data to uncover patterns in delivery performance across different routes, regions, and shipping modes. By quantifying lead times, identifying delay-prone routes, and evaluating geographic bottlenecks, the analysis aims to transform raw logistics data into actionable insights. The ultimate goal is to enable data-driven optimization of shipping operations and improve overall route efficiency.

## ANALYTICAL FRAMEWORK

This analysis follows a structured, KPI-driven approach to evaluate factory-to-customer shipping efficiency and identify operational bottlenecks.

* Data Validation & Cleaning: Ensure data integrity by standardizing date formats, removing invalid lead times, and handling missing records.

* Feature Engineering: Derive key variables such as Shipping Lead Time, route definitions (factory → region/state), and delay indicators.

* Route-Level Aggregation: Compute core metrics for each route, including shipment volume, average lead time, and variability.

* Performance Benchmarking: Rank routes to identify top-performing and underperforming corridors, and compare efficiency across ship modes.

* Geographic Analysis: Detect regions and states with high delays and congestion patterns.

* Operational Insights: Evaluate shipping modes and uncover cost–time trade-offs to support data-driven logistics optimization.

* Cross-Level Analysis: To check consistency translates from a higher route level to lower route levels.

This framework enables a transition from reactive monitoring to proactive, insight-driven decision-making in supply chain operations.

##KPI COMPUTATION
KPIs at route level- These metrics quantify delivery speed, reliability, and operational efficiency, enabling comparison across routes and shipping modes.

###Shipping Lead Time

```python
#Calculating Shipping Lead time
df_filtered['Shipping Lead Time'] = (df_filtered['Ship Date'] - df_filtered['Order Date']).dt.days
```

###Route-Level Aggregation : Average Lead Time & Route Volume

```python
#region level aggregation
route_agg_region = df_filtered.groupby('Route__Region').agg(Route_Volume = ('Shipping Lead Time', 'count'),Avg_Lead_Time=('Shipping Lead Time', 'mean'),Lead_Time_Variability=('Shipping Lead Time', 'std')).reset_index()
route_agg_region.head()
```

```python
#State level aggregation
route_agg_state = df_filtered.groupby('Route__State').agg(Route_Volume = ('Shipping Lead Time', 'count'), Avg_Lead_Time=('Shipping Lead Time', 'mean'),Lead_Time_Variability=('Shipping Lead Time', 'std')).reset_index()
route_agg_state.head()
```

###Delay Frequency

```python
#Delay Threshold
df_filtered['Shipping Lead Time'].describe()
```

```python
#Top 25% slowest shipments = delayed
Delay_threshold_region= df_filtered.groupby('Route__Region')['Shipping Lead Time'].quantile(0.75)
Delay_threshold_state= df_filtered.groupby('Route__State')['Shipping Lead Time'].quantile(0.75)
```

```python
df_filtered['Is_Delayed_region'] = df_filtered['Shipping Lead Time'] > df_filtered['Route__Region'].map(Delay_threshold_region)
df_filtered['Is_Delayed_state']= df_filtered['Shipping Lead Time'] > df_filtered['Route__State'].map(Delay_threshold_state)
```

```python
#Delay-Region Level
delay_region = df_filtered.groupby('Route__Region')['Is_Delayed_region'].mean().reset_index()
delay_region.rename(columns={'Is_Delayed_region': 'Delay_Frequency_RegionLevel'}, inplace=True)
delay_region['Delay_Frequency_RegionLevel'] = delay_region['Delay_Frequency_RegionLevel'] * 100
```

```python
#Delay-State Level
delay_state = df_filtered.groupby('Route__State')['Is_Delayed_state'].mean().reset_index()
delay_state.rename(columns={'Is_Delayed_state': 'Delay_Frequency_StateLevel'}, inplace=True)
delay_state['Delay_Frequency_StateLevel'] = delay_state['Delay_Frequency_StateLevel'] * 100
```

###Route Efficiency score

```python
#Region Level
region_avg = df_filtered.groupby('Route__Region')['Shipping Lead Time'].mean().reset_index()
region_avg['Normalized_Lead_Time']= ((region_avg['Shipping Lead Time'] - region_avg['Shipping Lead Time'].min()) / (region_avg['Shipping Lead Time'].max() - region_avg['Shipping Lead Time'].min()))
region_avg['Route_Efficiency_Score'] = 1 - region_avg['Normalized_Lead_Time']
```

```python
#State Level
state_avg = df_filtered.groupby('Route__State')['Shipping Lead Time'].mean().reset_index()
state_avg['Normalized_Lead_Time']= ((state_avg['Shipping Lead Time'] - state_avg['Shipping Lead Time'].min()) / (state_avg['Shipping Lead Time'].max() - state_avg['Shipping Lead Time'].min()))
state_avg['Route_Efficiency_Score'] = 1 - state_avg['Normalized_Lead_Time']
```

###KPI Table Construction

```python
#Region Level
Region_KPI_table = pd.merge(route_agg_region, delay_region, on='Route__Region', how='left')
Region_KPI_table = pd.merge(Region_KPI_table, region_avg, on='Route__Region', how='left')
Region_KPI_table = Region_KPI_table.drop(columns=['Shipping Lead Time'])
Region_KPI_table.head()
```

```python
#State Level
State_KPI_table = pd.merge(route_agg_state, delay_state, on='Route__State', how='left')
State_KPI_table = pd.merge(State_KPI_table, state_avg, on='Route__State', how='left')
State_KPI_table = State_KPI_table.drop(columns=['Shipping Lead Time'])
State_KPI_table.head()
```

##KEY INSIGHTS:
The analysis of shipping performance across region and state levels reveals important patterns in efficiency, bottlenecks, and operational consistency.

> 🚀 High Performing Routes

At the regional level, Canada → Atlantic emerges as the most efficient region, while Canada → Interior shows the weakest performance.

At the state level, top-performing routes include:

- United States → Atlantic → Maine
- Canada → Atlantic → Newfoundland And Labrador
- United States → Pacific → Nevada

In contrast, the least efficient routes are:

- United States → Atlantic → West Virginia
- United States → Interior → North Dakota
- Canada → Pacific → Saskatchewan

These results highlight that while certain regions perform well overall, performance can vary significantly across individual states.


> 🚨 Bottleneck Routes (High Volume + Poor Performance)

Routes combining high shipment volume with poor performance are identified as critical bottlenecks.

- At the regional level, United States → Atlantic stands out as a major bottleneck
- At the state level, key bottlenecks include:

  United States → Gulf → Georgia

  United States → Atlantic → New York

These routes indicate areas where high operational load may be contributing to inefficiencies and delays.


> ⚠️ Congestion-Prone Routes

Congestion-prone areas are identified based on high delay frequency and unreliable performance.

At the regional level,
- United States → Atlantic , shows consistent signs of congestion

At the state level,

- Canada → Pacific → Saskatchewan , emerges as a key congestion-prone route

These regions may require operational optimization to improve reliability.


> ❌ Worst-Case Routes (Slow + Delayed)

Routes that are both slow and delay-prone represent the most critical performance issues.

At the regional level,
- United States → Atlantic is identified as the worst-performing region

At the state level, the most critical routes include:

- Canada → Pacific → Saskatchewan
- United States → Interior → North Dakota

These routes require immediate attention due to combined inefficiencies in speed and reliability.

> 🔍 Cross-Level Insights

Cross-level analysis reveals that strong regional performance does not always translate to consistent state-level outcomes.

At the regional level, the most efficient regions are:
Atlantic (top)
Followed by Pacific and Gulf
However, at the state level, while United States → Atlantic → Maine is the top-performing route, aggregated performance trends differ:
Gulf emerges as the most consistently high-performing region
Followed by Atlantic

This indicates that regional averages can mask variability at the state level, emphasizing the importance of multi-level analysis.
